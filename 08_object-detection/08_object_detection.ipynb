```python
import cv2
from google.colab.patches import cv2_imshow
```

```python
%%bash
rm -rf /content/*
```

Sklonowanie repozytorium darknet

```python
%%bash
git clone https://github.com/pjreddie/darknet.git
cd darknet
make
```

Pobranie wag modelu YOLOv3

```python
%%bash
cd darknet
wget https://data.pjreddie.com/files/yolov3.weights
mkdir -p images
```

Wnioskowanie na podstawie modelu

```python
%%bash
cd darknet/
./darknet detect cfg/yolov3.cfg yolov3.weights images/office.jpg
```

Wyświetlanie predykcji

```python
img = cv2.imread('darknet/predictions.jpg')
cv2_imshow(img)
```
